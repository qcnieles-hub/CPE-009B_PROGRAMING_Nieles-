```python
%%writefile recipe.py
# General attributes and methods shared by all food types. OOP Concept used is encapsulation
class Recipe:
    def __init__(self, name, ingredients, instructions):
        self.name = name
        self.ingredients = ingredients
        self.instructions = instructions
        self.mealtype = "General"

    def display_recipe(self):
        print(f"\n--- {self.name.upper()} ({self.mealtype}) ---")
        print("Ingredients:")
        for ingredient in self.ingredients:
            print(f"- {ingredient.strip()}")
        print("Instructions:")
        print(self.instructions)

class Breakfast(Recipe):
   def __init__(self, name, ingredients, instructions):
    super().__init__(name, ingredients, instructions)
    self.mealtype = "Breakfast"


class Lunch(Recipe):
    def __init__(self, name, ingredients, instructions):
        super().__init__(name, ingredients, instructions)
        self.mealtype = "Lunch"


class Dinner(Recipe):
    def __init__(self, name, ingredients, instructions):
        super().__init__(name, ingredients, instructions)
        self.mealtype = "Dinner"

class Dessert(Recipe):
    def __init__(self, name, ingredients, instructions):
        super().__init__(name, ingredients, instructions)
        self.mealtype = "Dessert"

class Drinks(Recipe):
    def __init__(self, name, ingredients, instructions):
        super().__init__(name, ingredients, instructions)
        self.mealtype = "Drinks"
```

```python
%%writefile organizer.py
from recipe import Recipe, Breakfast, Lunch, Dinner, Dessert, Drinks

class RecipeOrganizer:
    def __init__(self):
        self.recipes = []
        self.load_recipes()

    def load_recipes(self):
        # Breakfast
        self.recipes.append(Breakfast("Beef Tapa", ["Beef", "Soy Sauce", "Calamansi", "Garlic"], "Marinate and pan fry."))
        self.recipes.append(Breakfast("Kimchi Rice", ["Cold Rice", "Kimchi", "Kimchi Juice", "Egg", "Sesame Oil"], "Fry kimchi and rice in skillet, top with fried egg."))
        self.recipes.append(Breakfast("Fish Dried with Coffee", ["Dried Fish (Tuyo)", "Cooking Oil", "Garlic Rice", "Hot Brewed Coffee", "Vinegar"], "Pan fry dried fish until crispy, serve with garlic rice and a hot cup of coffee with vinegar dip."))
        self.recipes.append(Breakfast("Spam, Fried Rice, and Sunny Side Up Egg", ["Spam Slices", "Cold Cooked Rice", "Eggs", "Garlic", "Cooking Oil"], "Pan fry spam slices until brown, saute garlic in oil and mix with cold rice to make garlic fried rice, and fry eggs sunny side up."))
        self.recipes.append(Breakfast("Pancit Canton with Egg and Rice", ["Pancit Canton Pack", "Boiled/Fried Egg", "Warm Cooked Rice", "Water"], "Boil noodles in water, drain, mix in seasoning packets, top with cooked egg, and serve alongside warm rice."))

        # Lunch
        self.recipes.append(Lunch("Chicken Adobo", ["Chicken", "Soy Sauce", "Vinegar", "Garlic"], "Simmer until soft."))
        self.recipes.append(Lunch("Tinulang Monggo", ["Mung Beans", "Pork", "Malunggay Leaves", "Garlic"], "Boil mung beans until soft, saute with garlic and pork, then add leaves."))
        self.recipes.append(Lunch("Lumpiang Gulay", ["Mixed Veggies", "Lumpia Wrapper", "Oil"], "Saute vegetables, wrap in lumpia wrappers, and deep fry."))
        self.recipes.append(Lunch("Pork Sinigang", ["Pork", "Tamarind", "Kangkong", "Radish"], "Boil pork until tender, add tamarind mix and vegetables."))
        self.recipes.append(Lunch("Pork Sisig", ["Pork Belly", "Onion", "Chili", "Calamansi"], "Boil, grill, chop pork, then saute with onions and chili."))

        # Dinner
        self.recipes.append(Dinner("Pork Menudo", ["Pork", "Liver", "Potatoes", "Carrots", "Tomato Sauce"], "Stew everything in tomato sauce."))
        self.recipes.append(Dinner("Garlic Butter Shrimp", ["Shrimp", "Butter", "Garlic", "Lemon"], "Sauté garlic in butter then cook shrimp."))
        self.recipes.append(Dinner("Egg with Malunggay Leaves Soup", ["Eggs", "Malunggay Leaves", "Garlic", "Onion", "Water"], "Saute garlic and onion, add water, drop eggs, and stir in malunggay leaves."))
        self.recipes.append(Dinner("Sinigang na Baboy", ["Pork", "Tamarind", "Kangkong", "Radish"], "Boil pork until tender, add tamarind and veggies."))

        # Dessert
        self.recipes.append(Dessert("Choco Chip Cookies", ["Flour", "Butter", "Sugar", "Choco Chips"], "Bake in oven."))
        self.recipes.append(Dessert("Mango Float", ["Graham", "Cream", "Condensed Milk", "Mango"], "Layer graham crackers and cream, then chill."))
        self.recipes.append(Dessert("Maja Blanca", ["Coconut Milk", "Cornstarch", "Sweet Corn", "Sugar", "Latik"], "Cook coconut milk with cornstarch and sweet corn until thick, transfer to mold, and top with latik."))
        self.recipes.append(Dessert("Buko Salad", ["Shredded Buko", "Fruit Cocktail", "Nata de Coco", "Nestle Cream", "Condensed Milk"], "Combine all ingredients in a bowl and chill thoroughly before serving."))

        # Drinks
        self.recipes.append(Drinks("Hot Chocolate with Marshmallows", ["Tablea / Cocoa Powder", "Milk", "Sugar", "Mini Marshmallows"], "Melt tablea or cocoa in warm milk, sweeten with sugar, pour into a mug, and float marshmallows on top."))
        self.recipes.append(Drinks("Iced Coffee with Condensed Milk", ["Brewed Coffee / Instant Coffee", "Condensed Milk", "Water", "Ice Cubes"], "Dissolve coffee in hot water, stir in condensed milk until smooth, and pour over a glass full of ice."))
        self.recipes.append(Drinks("Buko Juice with Scratched Buko", ["Fresh Young Coconut Water", "Scratched Buko Meat", "Sugar/Condensed Milk", "Ice"], "Mix coconut water with scraped young coconut meat, sweeten to taste, and chill with ice."))
        self.recipes.append(Drinks("Orange Juice", ["Fresh Oranges", "Water", "Sugar", "Ice Cubes"], "Squeeze oranges, mix juice with water and sugar to taste, then serve over ice."))

    def add_recipe(self):
        print("\n==== Adding a new recipe ====")
        foodname = input("Enter the food name: ")

        print("Select the meal type:")
        print("1. Breakfast")
        print("2. Lunch")
        print("3. Dinner")
        print("4. Dessert")
        print("5. Drinks")

        mealtype = input("Enter the meal type (1-5): ")
        ingredients = input("Enter the ingredients (comma-separated): ").split(",")
        instructions = input("Enter the instructions: ")

        if mealtype == "1":
            recipe = Breakfast(foodname, ingredients, instructions)
        elif mealtype == "2":
            recipe = Lunch(foodname, ingredients, instructions)
        elif mealtype == "3":
            recipe = Dinner(foodname, ingredients, instructions)
        elif mealtype == "4":
            recipe = Dessert(foodname, ingredients, instructions)
        elif mealtype == "5":
            recipe = Drinks(foodname, ingredients, instructions)
        else:
            print("Invalid selection.")
            return

        self.recipes.append(recipe)
        print(f"\nSuccessfully added '{foodname}'!")

    def display_recipes(self):
        if not self.recipes:
            print("No recipes found.")
            return

        print("\n==== Recipe List ====")
        for recipe in self.recipes:
            recipe.display_recipe()
```

```python
%%writefile NielesCenndy_Prelim.py
from organizer import RecipeOrganizer

# Main function to run the Recipe Organizer program.
def main():
    # Create an instance of RecipeOrganizer
    organizer = RecipeOrganizer()

    # Menu loop for user interaction
    while True:
        print("\n----------- Recipe Organizer Menu ---------")
        print("1. Add a recipe")
        print("2. View recipes")
        print("3. Exit")

        choice = input("Enter your choice (1-3): ")

        # Execute the corresponding action based on user input
        if choice == "1":
            organizer.add_recipe()
        elif choice == "2":
            organizer.display_recipes()
        elif choice == "3":
            print("Exiting the program.")
            break
        else:
            print("Invalid choice. Please try again.")

# Entry point of the program.
if __name__ == "__main__":
    main()
```
