```python
import random


# ============================================================
# Character (base class)
# ============================================================
class Character():
    def __init__(self, username):
        self.__username = username
        self.__hp = 100
        self.__mana = 100
        self.__damage = 5
        self.__str = 0
        self.__vit = 0
        self.__int = 0
        self.__agi = 0

    def getUsername(self):
        return self.__username

    def setUsername(self, new_username):
        self.__username = new_username

    def getHp(self):
        return self.__hp

    def setHp(self, new_hp):
        self.__hp = new_hp

    def getDamage(self):
        return self.__damage

    def setDamage(self, new_damage):
        self.__damage = new_damage

    def getStr(self):
        return self.__str

    def setStr(self, new_str):
        self.__str = new_str

    def getVit(self):
        return self.__vit

    def setVit(self, new_vit):
        self.__vit = new_vit

    def getInt(self):
        return self.__int

    def setInt(self, new_int):
        self.__int = new_int

    def getAgi(self):
        return self.__agi

    def setAgi(self, new_agi):
        self.__agi = new_agi

    def reduceHp(self, damage_amount):
        self.__hp = self.__hp - damage_amount

    def addHp(self, heal_amount):
        self.__hp = self.__hp + heal_amount


# ============================================================
# Novice (basic class, everyone starts here)
# ============================================================
class Novice(Character):
    def basicAttack(self, character):
        damage = self.getDamage() + 2
        character.reduceHp(damage)
        print(f"{self.getUsername()} performed Basic Attack! -{damage}")


# ============================================================
# Swordsman
# ============================================================
class Swordsman(Novice):
    def __init__(self, username):
        super().__init__(username)
        self.setStr(5)
        self.setVit(8)
        self.setHp(self.getHp() + self.getVit())

    def slashAttack(self, character):
        damage = self.getDamage() + self.getStr()
        character.reduceHp(damage)
        print(f"{self.getUsername()} performed Slash Attack! -{damage}")


# ============================================================
# Archer
# ============================================================
class Archer(Novice):
    def __init__(self, username):
        super().__init__(username)
        self.setAgi(7)
        self.setVit(4)
        self.setHp(self.getHp() + self.getVit())

    def rangedAttack(self, character):
        damage = self.getDamage() + self.getAgi()
        character.reduceHp(damage)
        print(f"{self.getUsername()} performed Ranged Attack! -{damage}")


# ============================================================
# Magician
# ============================================================
class Magician(Novice):
    def __init__(self, username):
        super().__init__(username)
        self.setInt(7)
        self.setVit(5)
        self.setHp(self.getHp() + self.getVit())

    def heal(self):
        self.addHp(5)
        print(f"{self.getUsername()} performed Heal! +5")

    def magicAttack(self, character):
        damage = self.getDamage() + self.getInt()
        character.reduceHp(damage)
        print(f"{self.getUsername()} performed Magic Attack! -{damage}")


# ============================================================
# Boss (combines all three advanced classes)
# ============================================================
class Boss(Swordsman, Archer, Magician):
    def __init__(self, username):
        super().__init__(username)
        self.setStr(5)
        self.setVit(10)
        self.setInt(5)
        self.setHp(self.getHp() + self.getVit())


# ============================================================
# Battle logic
# ============================================================
def battle(p1, p2):
    # random first turn
    if random.randint(0, 1) == 0:
        turn = 1
    else:
        turn = 2

    while p1.getHp() > 0 and p2.getHp() > 0:
        # PLAYER TURN
        if turn == 1:
            print("\nTurn:", p1.getUsername())
            if isinstance(p1, Swordsman):
                p1.slashAttack(p2)
            elif isinstance(p1, Archer):
                p1.rangedAttack(p2)
            elif isinstance(p1, Magician):
                p1.magicAttack(p2)
            else:
                p1.basicAttack(p2)
            turn = 2
        # ENEMY TURN
        else:
            print("\nTurn:", p2.getUsername())
            if isinstance(p2, Boss):
                p2.basicAttack(p1)
            elif isinstance(p2, Swordsman):
                p2.slashAttack(p1)
            elif isinstance(p2, Archer):
                p2.rangedAttack(p1)
            elif isinstance(p2, Magician):
                p2.magicAttack(p1)
            else:
                p2.basicAttack(p1)
            turn = 1

        # show hp
        print(p1.getUsername(), "HP:", p1.getHp())
        print(p2.getUsername(), "HP:", p2.getHp())

    # winner
    if p1.getHp() <= 0:
        return p2
    else:
        return p1


def chooseClass(name):
    print("\nChoose class:")
    print("1. Swordsman")
    print("2. Archer")
    print("3. Magician")
    choice = input("Enter: ")
    if choice == "1":
        return Swordsman(name)
    elif choice == "2":
        return Archer(name)
    else:
        return Magician(name)


def singlePlayer():
    wins = 0
    player = Novice("Player")
    player.setHp(120)  # slight advantage
    while True:
        boss = Boss("Monster")
        print("\n--- New Match ---")
        winner = battle(player, boss)
        print("Winner:", winner.getUsername())
        if winner.getUsername() == "Player":
            wins += 1
        print("Wins:", wins)

        # unlock class after 2 wins
        if wins == 2:
            player = chooseClass("Player")
            player.setHp(120)

        again = input("Play again? (y/n): ")
        if again != "y":
            break


# PLAYER VS PLAYER
def playerVsPlayer():
    name1 = input("Player 1 name: ")
    name2 = input("Player 2 name: ")
    p1 = chooseClass(name1)
    p2 = chooseClass(name2)
    while True:
        print("\n--- New Match ---")
        winner = battle(p1, p2)
        print("Winner:", winner.getUsername())
        p1.setHp(100)
        p2.setHp(100)
        again = input("Play again? (y/n): ")
        if again != "y":
            break


# MAIN MENU
def main():
    print("GAME MENU")
    print("1. Single Player")
    print("2. Player vs Player")
    mode = input("Choose: ")
    if mode == "1":
        singlePlayer()
    elif mode == "2":
        playerVsPlayer()
    else:
        print("Invalid choice")


if __name__ == "__main__":
    main()
```
